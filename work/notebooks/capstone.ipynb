## 5-Minute Demo Outline (Week-8 Showcase)

**Question (30s):** Which pages should a content team with limited review
capacity check first, out of thousands?

**Method (60s):** Trained a Random Forest on first-half-of-month observable
signals (impressions, position, age, CTR) from FlyRank's 79M-row production
search warehouse, validated against a transparent hand-written baseline rule
under a client-grouped split — no client's pages appear in both train and test.

**One chart (60s):** Results table — Baseline 0.560 vs. Random Forest 0.740
Precision@50, both on the same held-out clients.

**One honest result (90s):** The real finding isn't the model beating the
baseline — it's that scoring the *same model* on a naive, ungrouped split
would have shown 0.940, a 0.200 overstatement caused by 44 clients leaking
across train and test. How you validate matters as much as what you build.

**One recommendation (60s):** Of 14,520 test pages, only 143 get flagged for
direct human review — a workload sized to a real team's capacity, with an
explicit rule that nothing here auto-publishes, auto-emails, or auto-removes
anything without a human checking it first.
